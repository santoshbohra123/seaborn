heat map - > plot rectangular data as acolour- encoded matrix.


```python
import seaborn as sns
import plotly.express as px
import matplotlib.pyplot as plt
```

```python

tips = sns.load_dataset('tips')
gap = px.data.gapminder()
```

```python
df = gap.pivot(index = 'country', columns= 'year' ,values = 'lifeExp', )
```

```python
# axes level function
# not exist figure level function 

plt.figure(figsize = (15,12))
sns.heatmap(df)
```

```python
temp_df = gap[gap['continent'] == 'Europe'].pivot(index = 'country', columns = 'year', values = 'lifeExp')
# print(temp_df)

plt.figure(figsize = (15,12))
sns.heatmap(temp_df,annot = True,cmap = 'viridis',linewidth = 0.2)

#annot = True -> give exact values
```
