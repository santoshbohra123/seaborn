```python
import seaborn as sns
import plotly.express as px
```

```python
# tells counts of value using bins

# here also are figure level and axes level function

# fig lvel func name here -> distplot()
# axes level func ->  histplot(), kdeplot(), rugplot
```

```python
# Histplot

tips = sns.load_dataset('tips')
gap = px.data.gapminder()
```

```python
sns.histplot(data=tips ,x = 'total_bill')
```

```python
sns.displot(kind = 'hist', data = tips, x = 'total_bill')
```

```python

# no of bins params
sns.displot(kind = 'hist', data = tips, x = 'total_bill',bins = 10)


```

```python
# it is alsoo possible the visualise histogram on categorical variable. discreate bins automatically set for categorical vriable.
sns.displot(kind = 'hist', data = tips, x = 'day',)

# It is acting like a countplot

```

```python
# hue parameter
sns.displot(kind = 'hist', data = tips, x = 'total_bill',bins = 10,hue  = 'sex')
```

```python
# element -> step
sns.displot(kind = 'hist', data = tips, x = 'total_bill',bins = 10,element = 'step',hue = 'smoker')
```

```python
titanic = sns.load_dataset('titanic')
titanic
```

# KDE Plot -> Kernel density estimation 

 rather than using discreate bins, a KDE plot smooths the observations with the gausian kernel.
 it produce continuous density estimates


```python
tips
```

```python
sns.kdeplot(data = tips, x = 'total_bill')
```

```python
# it shows density or we can say probability for understanding of total bills.
# it shows 20 has more density in a dataset.
# not any details miss in continuous. 
```

```python
sns.displot(kind = 'kde', data = tips, x = 'total_bill', hue = 'sex', fill = True)

# fill = True it tells the area .
```

```python
# RUG plot

# plot marginal distribution by drawing ticks along x and y axis 
# show same info in different form.


sns.kdeplot(data = tips, x = 'total_bill',)

sns.rugplot(data = tips, x = 'total_bill')
```
