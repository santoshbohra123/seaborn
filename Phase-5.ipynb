# Regression plot

regplot
lmplot
both function draw the sctter plot of two variables,x and y , and then fit the regression model y~x and plot result as regression line 

```python
import seaborn as sns
```

```python
tips = sns.load_dataset('tips')
iris =  sns.load_dataset('iris')
```

```python
# hue params is not available ->  hue = 'sex' 

sns.regplot(data=tips, y = 'tip', x='total_bill',)
```

```python
sns.lmplot(data=tips, y = 'tip', x='total_bill',hue='sex')
```

```python
# residual plot residplot()


sns.residplot(data=tips, x = 'total_bill', y = 'tip')

```
