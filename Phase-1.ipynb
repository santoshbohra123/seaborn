```python
import seaborn as sns
import matplotlib.pyplot as plt
import plotly.express as px
```

```python
tips = sns.load_dataset('tips')
tips.head()
```

```python
# Scatter plot -> Axes level function

sns.scatterplot(data = tips,x = 'total_bill', y = 'tip' , hue = 'sex', style = 'time',size = 'size')

# using hue parameter we can easily classify dots based on columns. it give legend () fn of matplotlib automatically.
# style = 'time' -> it tells another filter based on time.
#  size = 'size' parameter btata h koi customer kitne logo ke sath aya jo jitne jyada logo ke sath aya uska size utna hi bda hoga.
# provide added information using hue, size,style params.
```

```python
# rel plot is figure level function -> generally square shape default.

sns.relplot(data = tips, x = 'total_bill', y = 'tip', kind= 'scatter',hue = 'smoker',style = 'time',size = 'size')

#  hue =  Change color
#  size =  maker shape change krta h
#  shape = maker size change krta h
```

```python
#        # Line plot

# if x axis has time than it is usually good to see in line plot
```

```python
gap = px.data.gapminder()
gap
```

```python
india = gap[gap['country'] == "India"]
india
```

```python
# axes level function
sns.lineplot(data = india, x = 'year', y = 'lifeExp')
```

```python
country_df = gap[gap['country'].isin(["India", 'China', 'Japan'])]
country_df.head(15)
```

```python
# Figure level function
#  it has more power or control than axes so it can clearly take outside from graph the legend() 
# while axes function provide legend() inside the plot.

# size - thikness of a line  

sns.relplot(data = country_df, x = 'year', y = 'lifeExp',kind = 'line',hue = 'country')
```

facet plot

```python
# facet plot only work over relplot() or fig level functions because they have property to control figure using row='' , col='' params.
# axes level functions like scatterplot, linepot has not a have this property.
sns.relplot(data = tips, x = 'total_bill', y = 'tip', kind = 'scatter', col = 'smoker', row = 'sex')
```

```python
# col_wrap
sns.relplot(kind = 'scatter', data = gap, x = 'gdpPercap', y= 'lifeExp',col = 'year',col_wrap = 3,hue = 'country')
```
