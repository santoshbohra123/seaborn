```python
import seaborn as sns
```

```python
tips = sns.load_dataset('tips')
iris =  sns.load_dataset('iris')
```

# MultiGridplot

facetgrid
jointplot
pairplot

```python
# Facet grid 


# another way to use facet cols and row in graphs

g = sns.FacetGrid(data=tips, col='day')
# g.add_legend()
g.map(sns.boxplot,'tip','total_bill')
```

```python
#  Pair plot                                    imp

# automatically identify neumerical columns and make scatterplot on every possible pair of these columns in dataset.
# with same column it plot histogram and for different plot scatter plot.
# it form a symmetric matric -> diagonal above and below have same graphs exist.


sns.pairplot(iris)

```

```python
sns.pairplot(iris, hue='species')
```

```python
# adding hue parameter it produce Kde plot at diagonals.
```

```python
# plotting pairwise relationship (pairgrid vs pairplot)

# pairplot forms by pairGrid


g = sns.PairGrid(data=iris,hue = 'species')
g.map(sns.scatterplot)  # this plot all scatter plots .
```

```python
g  = sns.PairGrid(data=iris, hue='species')
g.map_diag(sns.kdeplot)
g.map_offdiag(sns.kdeplot)


# we can customised these plots using self.
```

```python
# customised diagonal upper and lower 

g  = sns.PairGrid(data=iris, hue='species')
g.map_diag(sns.histplot)
g.map_upper(sns.kdeplot)
g.map_lower(sns.scatterplot)

# Pair plot is only one type of pair grid

```

```python
 #  vars  -> customised neumerical columns of data

g  = sns.PairGrid(data=iris, hue='species',vars=['sepal_width', 'petal_width'])
g.map_diag(sns.histplot)
g.map_offdiag(sns.scatterplot)

```

# Joint plot and joint grid


```python
# focus on 2 variable and one variable simultainiously.
# study of 2d and 1d simultainiously.



sns.jointplot(data=tips, x = 'total_bill', y = 'tip',kind = 'kde',hue='sex')

# kind = 'scatter' it is default  -->  change graph type. 
# kind = 'hist' 2d and histogram
# kind = 'reg'  -> kde +histogram , and reg plot
#kind = 'kde'  -> 2d kde and kdeplot
```

```python
# Joint grid
```

```python
g =sns.JointGrid(data=tips,x = 'total_bill', y = 'tip')
g.plot(sns.scatterplot,sns.boxplot)
```
