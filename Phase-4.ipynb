```python
import seaborn as sns
```

```python
tips = sns.load_dataset('tips')
iris =  sns.load_dataset('iris')
tips

```

# Categorical Plots

    categorical scatter plots
       
    Categorical distribution plots

    categorical Estimate plot -> for central tendency

```python
# stripplot() -> use for categorical vs neumerical dataset. 
# it is a type of scatter plot.

# axes level function 

sns.stripplot(data= tips, x= 'day' , y = 'total_bill', jitter= False)
```

```python
# figure level function 
sns.catplot(kind = 'strip', data = tips , x = 'day', y = 'total_bill')
```

```python
# jitter
# jitter adds noise to view perfactly
sns.catplot(kind = 'strip', data = tips , x = 'day', y = 'total_bill',jitter = 0.2, hue = 'sex')
```

```python
# swarmplot - > Data distribution has good understandabilty over strip plot.
# useful for small dataset
# using figure level function

sns.catplot(data=tips ,kind = 'swarm' ,x='day' , y = 'total_bill')
```

```python
sns.swarmplot(data=tips , x='day' , y = 'total_bill',hue = 'day')
```
