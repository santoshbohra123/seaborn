Categorical plots

```python
import seaborn as sns
```

```python
tips = sns.load_dataset('tips')
iris =  sns.load_dataset('iris')
tips

```

# Categorical Plots

    categorical scatter plots
       
    Categorical distribution plots

    categorical Estimate plot -> for central tendency

```python
# stripplot() -> use for categorical vs neumerical dataset. 
# it is a type of scatter plot.

# axes level function 

sns.stripplot(data= tips, x= 'day' , y = 'total_bill', jitter= False)
```

```python
# figure level function 
sns.catplot(kind = 'strip', data = tips , x = 'day', y = 'total_bill')
```

```python
# jitter
# jitter adds noise to view perfactly
sns.catplot(kind = 'strip', data = tips , x = 'day', y = 'total_bill',jitter = 0.2, hue = 'sex')
```

```python
# swarmplot - > Data distribution has good understandabilty over strip plot.
# useful for small dataset
# using figure level function

sns.catplot(data=tips ,kind = 'swarm' ,x='day' , y = 'total_bill')
```

```python
sns.swarmplot(data=tips , x='day' , y = 'total_bill',hue = 'day')
```

# Categorical Distribution plot

1 - boxplot
2 - violinplot


```python
# Box plot

# it is a standardise way of displaying the distribution of data based on 5 numbers summary(min,q1,median,q3,max). 
# tell about  outliers and remove them . 
# tells  skewness and symmetry

```

```python
# using axes 

sns.boxplot(data = tips, y = 'total_bill')
```

```python
# usinf figure level
sns.catplot(data = tips, x = 'day',  y = 'total_bill',kind = 'box', hue= 'sex')
```

```python
# single box plot
sns.boxplot(data = tips, y = 'total_bill')
```

# violin plot = (boxplot + kdeplot)



```python
sns.violinplot(data = tips , x= 'day', y = 'total_bill')  # axes
```

```python
sns.catplot(kind = 'violin',data = tips , x= 'day', y = 'total_bill', hue='sex')
```

```python
sns.catplot(kind = 'violin',data = tips , x= 'day', y = 'total_bill', hue='sex',split = True)
```

```python
# Box plot is more useful than violin plot
```

# Categorical Estimate plot

# bar plot



```python
import numpy as np
```

```python
sns.barplot(data = tips, x = 'sex', y = 'total_bill',hue = 'smoker',estimator  = np.std)

# it default calculate mean we can change its value like std.
```

```python
# when there are multiple observtion in each category , it also uses bootstraping to compute a confidence interval around the estimate which is plotted using error br.

sns.catplot(kind = 'bar', data = tips, x = 'sex', y = 'total_bill',hue = 'smoker',estimator  = np.mean)
```

```python
sns.barplot(data = tips, x = 'sex', y = 'total_bill')

#  we calculate statistics(mean , meadian ,mode, min, max etc.) using categorical and neumerical variable. 
```

```python
#

#   point plot

sns.pointplot(data = tips, x = 'sex', y = 'total_bill')

# it connects error bars.
#  point plot = barplot - bars from it
# it calculates differece between categoriess.

#  we calculate difference using categorical and neumerical variable. 


```

```python
sns.pointplot(data = tips, x = 'sex', y = 'total_bill', hue='smoker')
sns.catplot(kind='bar',data = tips, x = 'sex', y = 'total_bill', hue='smoker')
```

```python
# count plot
# we calculate  only count statistics of categories.

sns.countplot(data = tips, x = 'sex', hue='day')

# histogram does same work for neumerical data.
```

```python
# faceting using catplot


# default value of catplot is strip plot.
sns.catplot(data  = tips , x = 'sex', y = 'total_bill', col='smoker', kind='box',row='time')

# it has also property of col_wrap
```
